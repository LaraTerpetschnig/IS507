```python
import pandas as pd
```

```python
#reading the csv file into a pandas dataframe
df = pd.read_csv("UIC Police Daily Crime-Fire Log 07 Sep 26.csv")
df
```

```python
#dropping the last column of the dataframe since it is empty
df = df.drop(columns=df.columns[7]) 
df
```

```python
#renaming the column 'UICPD\nReport #' to 'UICPD Report Number' for better readability
df = df.rename({'UICPD\nReport #':'UICPD Report Number'}, axis=1)
df
```

```python
#splitting the dataframe into two separate dataframes, one for criminal incidents and one for non-criminal incidents
split_num = df[df['Incident'].str.startswith('Non-Criminal Incidents')].index[0]
split_num
```

```python
#creating a new dataframe for criminal incidents
df_crime = df.iloc[:split_num]
df_crime
```

```python
#dropping the first three rows and the last two rows from the criminal incidents dataframe
#doing this because these rows were NA values that separated the 2 tables and not used for any data information
df_crime = df_crime.drop([0, 1, 2]) 
df_crime = df_crime.drop(df_crime.tail(2).index)  
df_crime
```

```python
#adding a new column to the criminal incidents dataframe called 'Incident Type' and setting all values to 'Criminal/Fire'
df_crime['Incident Type'] = "Criminal/Fire" 
df_crime
```

```python
#creating a new dataframe for non-criminal incidents
df_not_crime = df.iloc[split_num:]
df_not_crime
```

```python
#repeating dropping the NA rows for the non-criminal incidents dataframe
df_not_crime = df_not_crime.drop(df_not_crime.head(2).index) # drop first 2 rows 
df_not_crime
```

```python
#repeating adding a incident type column and labeling as non-criminal
df_not_crime['Incident Type'] = 'Non-Criminal'
df_not_crime
```

```python
#combining the two dataframes into one
df_combine = pd.concat([df_crime, df_not_crime])
df_combine
```

```python
#splitting the 'Location / Bldg #' column into three separate columns: 'Location', 'Building', and 'Building Number'
df_combine[['Location', 'Building', 'Building Number']] = df_combine['Location / Bldg #'].str.split('/', expand=True)
df_combine = df_combine.drop('Location / Bldg #', axis=1)
df_combine
```

```python
#cleaning column names for better readability
df_combine['Building Number'] = df_combine['Building Number'].str.replace('Building', '')
df_combine['Building Number'] = df_combine['Building Number'].str.replace('\n', '')
df_combine
```

```python
#cleaning the 'Date/Time Reported' column to remove any newline characters
df_combine['Date/Time Reported'] = df_combine['Date/Time Reported'].str.replace('/\n', '')
df_combine
```

```python
#converting the date in reported column to number only format
df_combine['Date/Time Reported'] = pd.to_datetime(df_combine['Date/Time Reported'], format='%d %b %y %I:%M %p')
df_combine
```

```python
#cleaning the 'Date/Time Occurred' column to remove any newline characters
df_combine['Date/Time Occurred'] = df_combine['Date/Time Occurred'].str.replace('/\n', '')
df_combine['Date/Time Occurred'] = df_combine['Date/Time Occurred'].str.replace('\n', '')
df_combine
```

```python
#splitting the occured column into start and end columns to compare end time with repoted time later
df_combine[['Date/Time Occurred Start', 'Date/Time Occurred End']] = df_combine['Date/Time Occurred'].str.split('-', expand=True)
df_combine = df_combine.drop('Date/Time Occurred', axis=1)
df_combine
```

```python
#adding the date to the end column

df_combine['Date/Time Occurred End'] = df_combine['Date/Time Occurred Start'].apply(lambda x : ' '.join(x.split()[:3])) + ' ' + df_combine['Date/Time Occurred End']
df_combine
```

```python
#converting the start and end columns to datetime format 
df_combine['Date/Time Occurred Start'] = pd.to_datetime(df_combine['Date/Time Occurred Start'], format='mixed')
df_combine['Date/Time Occurred End'] = pd.to_datetime(df_combine['Date/Time Occurred End'], format='mixed')
df_combine
```

```python
#rnaming the comments column to something shorter
df_combine = df_combine.rename({'Comments / Nature of Fire':'Comments'}, axis=1)
df_combine
```
